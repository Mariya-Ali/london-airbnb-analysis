```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10,8))
plt.scatter(df["longitude"], df["latitude"], c=df["price"], cmap="viridis", alpha=0.5, s=5, vmax=300)
plt.colorbar(label="Price (£)")
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("Airbnb Price Distribution in London")
plt.show()


```

```python
df["room_type"].value_counts().plot(kind="bar")
plt.title("Room Type Distribution")
plt.xlabel("Room Type")
plt.ylabel("Count")
plt.show()
```

```python
df["price"].hist(bins=50, range=(0,500))
plt.title("Price Distribution")
plt.xlabel("Price (£)")
plt.ylabel("Frequency")
plt.show()
```

```python
import matplotlib.pyplot as plt
df.boxplot(column="price", by="room_type", figsize=(8,6))
plt.title("Price Distribution by Room Type")
plt.suptitle("")
plt.ylabel("Price (£)")
plt.ylim(0, 500)
plt.show()
```
